```python
import pandas as pd
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
df.head()
```

```python
df.info()
```

```python
df.describe(include ="all")
```

```python
df.isnull().sum()
```

```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.mean()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df 
```

```python
df = df.rename(columns = {'purchase_amount_(usd)': 'purchase_amount'})
df.columns
```

```python
### Feature Engineering
```

```python
# Creating a column age_group
labels = ["Young Adult","Adult","Middle-aged","Senior"]
df["age_group"] = pd.qcut(df["age"],q = 4,labels=labels)
```

```python
df[["age",'age_group']].head(10)
```

```python
# creating column Purchase _frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Dropping promo code used column since it is same as discount applied
df = df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

```python
df.to_csv("customer.csv", index=False)
```
